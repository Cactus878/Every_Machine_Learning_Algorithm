```python
import pandas as pd
import re
```

```python
# P(A|B)= P(B|A) * P(A) / P(B)
```

```python
df = pd.read_csv('emails.csv')
spam_df = df[df['spam'] == 1]
real_df = df[df['spam'] == 0]
```

```python
def clean(tokens):
    new_tokens = []
    for i in tokens:
        text = str(i).lower()

        text = re.sub(r'http\S+|www\S+|https\S+', '', text)
        text = re.sub(r'@\w+|#', '', text)
        text = re.sub(r'[^a-z\s]', '', text)
        text = re.sub(r'\s+', ' ', text).strip()
        text = re.sub(r'\brt\b', ' ', text).strip()
        
        if text in STOP_WORDS:
            text = ""
            
        if len(text) < 2:
            text = ""

        if text:
            new_tokens.append(text)
            
    if new_tokens[0] == 'subject':
        new_tokens.pop(0)

    return new_tokens
```

```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
nlp = spacy.load("en_core_web_sm")
```

```python
spam_email_tokens = []
real_email_tokens = []

for _, i in spam_df.iterrows():
    tokens = []
    email = nlp(i[0])
    for j in email:
        tokens.append(j)
    spam_email_tokens.append(tokens)

for _, i in real_df.iterrows():
    tokens = []
    email = nlp(i[0])
    for j in email:
        tokens.append(j)
    real_email_tokens.append(tokens)
```

```python
cleaned_spam_email_tokens = [clean(x) for x in spam_email_tokens]
cleaned_real_email_tokens = [clean(x) for x in real_email_tokens]
```

```python
spam_email_words = pd.DataFrame({"Word": [], "Count": []})
real_email_words = pd.DataFrame({"Word": [], "Count": []})
```

```python
def countWords(df, email_tokens):
    for email in email_tokens:
        for token in email:
            if token in df["Word"].values:
                df.loc[df["Word"] == token, "Count"] += 1
            else:
                df.loc[len(df)] = [token, 1]
    return
```

```python
countWords(spam_email_words, cleaned_spam_email_tokens)
countWords(real_email_words, cleaned_real_email_tokens)
```

```python
spam_email_words.sort_values(by='Count', ascending=False)
```

```python
real_email_words.sort_values(by='Count', ascending=False)
```

```python
def calculateProbability(df):
    sum_amount_of_words = df['Count'].sum()
    df['Probability'] = df['Count'] / sum_amount_of_words
    return df
```

```python
spam_email_probabilities = calculateProbability(spam_email_words)
real_email_probabilities = calculateProbability(real_email_words)
```

```python
spam_email_probabilities
```

```python
real_email_probabilities
```

```python
def predictEmail(email, real_probabilites, spam_probabilites, real_email_count, spam_email_count):
    words_of_email = nlp(email.lower())

    spam_score = 1
    real_score = 1

    spam_scores = [spam_email_count / (real_email_count+spam_email_count)]
    real_scores = [real_email_count / (real_email_count+spam_email_count)]
    
    for word in words_of_email:
        if word.text in spam_probabilites['Word'].values:
            spam_scores.append(spam_probabilites.loc[spam_probabilites["Word"] == word.text, "Probability"].iloc[0])
            
    for word in words_of_email:
        if word.text in real_probabilites['Word'].values:
            real_scores.append(real_probabilites.loc[real_probabilites["Word"] == word.text, "Probability"].iloc[0])

    for score in spam_scores:
        spam_score = spam_score * score

    for score in real_scores:
        real_score = real_score * score

    if spam_score > real_score:
        print(f"With a value of {spam_score}, the email has been identified as spam")
    elif spam_score < real_score:
        print(f"With a value of {real_score}, the email has been identified as real")
    else:
        print(f"Undecided")
        
    return spam_score, real_score
```

```python
spam_email_count = len(spam_df)
real_email_count = len(real_df)
```

```python
spam, real = predictEmail("Click here for millions of money", real_email_probabilities, spam_email_probabilities, real_email_count, spam_email_count)
```
